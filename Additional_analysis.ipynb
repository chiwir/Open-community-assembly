```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# Data
J_values = np.array([1000, 3000, 5000, 7000, 9000, 
                     11000, 13000, 15000, 17000, 19000])
avg_richness = np.array([21, 44, 66, 80, 93, 
                          104, 115, 122, 131, 138])

# Linear regression
slope, intercept, r_value, p_value, std_err = linregress(J_values, avg_richness)

# Predicted values
y_pred = slope * J_values + intercept

# Print regression results
print(f"Slope: {slope:.5f}")
print(f"Intercept: {intercept:.5f}")
print(f"R-squared: {r_value**2:.4f}")
print(f"P-value: {p_value:.4e}")

# Plot
plt.figure()
plt.scatter(J_values, avg_richness, label="Observed data")
plt.plot(J_values, y_pred, label="Linear fit")
plt.xlabel("J values")
plt.ylabel("Mean richness")
plt.legend()
plt.show()
```
